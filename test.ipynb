#### Without pydantic

```python
def insert_patient_data(name: str, age: int):
    if type(name) == str and type(age) == int:
        if age > 0:
            print(f"Name: {name}")
            print(f"Age: {age}")
            print("Data Added to the Database!")
        else:
            raise ValueError("Age cannot be negative")
    else:
        raise TypeError("Name should be string & Age should be a number")

insert_patient_data("Alex", 25)
```

## With Pydantic

```python
from pydantic import BaseModel
```

```python
class PatientData(BaseModel):
    name: str
    age: int

def insert_patient_data(pateint: PatientData):
    print(f"Name: {pateint.name}")
    print(f"Age: {pateint.age}")
    print("Data added sucessfully to the database")

patient_data = {"name": "Alex", "age": "24"}

patient_1 = PatientData(**patient_data)
insert_patient_data(patient_1)

```

### A little advanced

```python
from typing import List, Dict
'''
Older python version didnt support List[str] & Dict[str, str] so we using 'typing' library
'''
class PatientData(BaseModel):
    name: str
    age: int
    married: bool
    allergies: List[str]
    contact_info: Dict[str, str]


def insert_patient_data(patient: PatientData):
    print(f"Name: {patient.name}")
    print(f"Age: {patient.age}")
    print(f"Married: {patient.married}")
    print(f"Allergies: {patient.allergies}")
    print(f"contact information: {patient.contact_info}")
    print("Data added sucessfully to the database")

patient_data = {"name": "Alex", "age": "24", "married": "True", "allergies": ["dust", "milk"], "contact_info": {"Email": "xyz@gmail.com", "ContactNo": "00000000"}}

patient_1 = PatientData(**patient_data)
insert_patient_data(patient_1)
```

### Required & Optional fields in the Schema

```python
from typing import List, Dict, Optional
'''
import "Optional" incase a user does not enter a value, so this will not raise an error
and to set a default value for a variable you can simply add an assignemnt operator after the data type
'''
class PatientData(BaseModel):
    name: str
    age: int
    married: bool = False
    allergies: Optional[List[str]]
    contact_info: Dict[str, str]


def insert_patient_data(patient: PatientData):
    print(f"Name: {patient.name}")
    print(f"Age: {patient.age}")
    print(f"Married: {patient.married}")
    print(f"Allergies: {patient.allergies}")
    print(f"contact information: {patient.contact_info}")
    print("Data added sucessfully to the database")

patient_data = {"name": "Alex", "age": "24", "allergies": ["dust", "milk"], "contact_info": {"Email": "xyz@gmail.com", "ContactNo": "00000000"}}

patient_1 = PatientData(**patient_data)
insert_patient_data(patient_1)
```

## Data Validation

```python
from pydantic import BaseModel, EmailStr, AnyUrl, Field
from typing import List, Dict, Optional
'''
Even though a field is specified a "str" type, it can also can issue if a user tries to add a long line of paragraphs
for ex: a field specified for "email" should be string, and normally the email consist of 20-25 chars, but what if the user enters an email which is 1000 chars long?
that would be wrong right? so with pydantic one can specify the length of the input for a specific variable!

'''
class PatientData(BaseModel):
    name: str = Field(max_length=50)
    email: EmailStr
    linkdln_url: Optional[AnyUrl]
    age: int = Field(gt=0, lt=100) # gt = greater than, lt = lesser than


def print_data(patient: PatientData):
    print(f"Name: {patient.name}")
    print(f"Email: {patient.email}")
    print(f"Linkedln: {patient.linkdln_url}")
    print(f"Age: {patient.age}")
    print("Data Printed sucessfully!")


patient_data = {"name": "Bob", "email": "xyz@gmail.com", "linkdln_url": "https://www.linkedin.com/in/mushtaq-saeed/", "age": "20"}

patient_2 = PatientData(**patient_data)
print_data(patient_2)
```

### Field attribute 



```python
from pydantic import BaseModel, EmailStr, AnyUrl, Field
from typing import List, Dict, Optional, Annotated
'''
We need to use Annotated object to clarify a field more clearly, also we can additional information and restriction, for example: what type of data should exactly a variable should take

'''
class PatientData(BaseModel):
    name: Annotated[str, Field(max_length=50, title='Name of the patient', description='Give the name of the patient in less than 50 chars', examples=['Alex', "clara"])]
    email: EmailStr
    linkdln_url: Optional[AnyUrl]
    age: Annotated[Optional[int], Field(gt=0, lt=100, strict=True)] = None # gt = greater than, lt = lesser than / Strict means that value must be exactly an integer or whatever was specified


def print_data(patient: PatientData):
    print(f"Name: {patient.name}")
    print(f"Email: {patient.email}")
    print(f"Linkedln: {patient.linkdln_url}")
    print(f"Age: {patient.age}")
    print("Data Printed sucessfully!")


patient_data = {"name": "Bob", "email": "xyz@gmail.com", "linkdln_url": "https://www.linkedin.com/in/mushtaq-saeed/", "age": 20}

patient_2 = PatientData(**patient_data)
print_data(patient_2)
```
